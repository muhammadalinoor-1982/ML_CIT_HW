```python
# Armstrong Calculation
n = '153'
sum1 = 0
for i in n:
    print(int(i)**3)
    sum1 = sum1 + int(i)**3
sum1
```

```python
# Prime Number
n = 8
p = 'prime number'
for i in range(2,n):
    if n%i==0:
        print(i)
        p = 'not a prime number'
        break
print(p)
```

```python
# Fibonacci Series

seq = int(input('Enter the range '))
a = 0
b = 1
if seq == 0:
    list1 = []
elif seq == 1:
    list1 = [0]
elif seq >= 2:
    list1 = [0,1]
    for i in range(seq-2):
        c = a+b
        a = b
        b = c
        list1.append(c)
list1
```
