```python
df = pd.read_csv('iris.csv')
df
```

```python
df['species'].value_counts()
```

```python
df.plot(kind='scatter', x='sepal_length', y='sepal_width')
plt.show()
```

```python
sns.set_style('whitegrid')
sns.FacetGrid(df, hue='species').map(plt.scatter, 'sepal_length', 'sepal_width').add_legend()
plt.show()
```

```python
df = px.data.iris()
fig = px.scatter_3d(df, x='sepal_length', y='sepal_width', z='petal_width', color='species')
fig.show()
```

```python
sns.set_style('whitegrid')
sns.pairplot(df[['sepal_length', 'sepal_width', 'petal_length', 'petal_width', 'species']], hue='species', height=3)
plt.show()
```

```python
sns.set_style('whitegrid')
sns.pairplot(df, hue='species', height=3)
plt.show()
```

```python
sns.boxplot(x='species', y='petal_length', data=df, hue='species')
plt.show()
```

```python
sns.violinplot(x='species', y='petal_length', data=df, hue='species')
plt.show()
```
